```python
import pandas as pd
import plotly.express as px
```

```python
df_progression = pd.read_csv("/workspaces/K-Means/data/progression.csv")
df_progression.sample(5)
```

```python
df_progression["lang_version"] = df_progression["language"] + "-" + df_progression["version"]

metrics = [
    "total_time_sec",
    "throughput_exec_per_sec",
    "avg_latency_ms"
]

for metric in metrics:
    fig = px.line(
        df_progression,
        x="threads",
        y=metric,
        color="lang_version",
        markers=True,
        title=f"Threads vs {metric}",
    )

    fig.update_layout(
        xaxis_title="Número de Threads",
        yaxis_title=metric,
        template="plotly_dark",
        hovermode="x unified"
    )

    fig.show()
```
